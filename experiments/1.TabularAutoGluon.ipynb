## Fit Tab AutoGluon

```python
from autogluon.tabular import TabularDataset, TabularPredictor
import pandas as pd

# Load the dictionary saved in the pickle file.
setup = pd.read_pickle("../data/cv_setup.pkl")

# Extract the two pandas DataFrames and the list of CV splits.
train = setup["train_data"]
validation = setup["validation_data"]
cv_splits = setup["cv_splits"]

# target name
label = "spread"

# train the model
predictor = TabularPredictor(
    label=label,
    # eval_metric=metric # somehow specify the customer cumulative PnL metric
).fit(
    train,
    time_limit=60 # stops training after x seconds, use Kaggle limit - saving time
)
```

## Performance

```python
# make a prediction on the validation data
y_pred = predictor.predict(validation.drop(columns=[label]))

# evaluate the performance on the validation data
predictor.evaluate(validation)

# evaluate the performance of each trained model on the validation data
predictor.leaderboard(validation)

# estimate the overall importance of features
# importance estimates evaluation metric drop if the feature were removed from the data (removal of negatives likely improves the eval metric)
predictor.feature_importance(validation)
```
